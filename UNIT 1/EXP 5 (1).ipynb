```python
# Experiment: Text Generation Using a Pre-trained LLM and Response Analysis

# Step 1: Install required libraries
!pip install -q transformers torch vaderSentiment

# Step 2: Import libraries
from transformers import pipeline
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
import pandas as pd

print("=" * 60)
print("TEXT GENERATION AND RESPONSE ANALYSIS USING GPT-2")
print("=" * 60)

# Step 3: Load the pre-trained GPT-2 model
print("\nLoading pre-trained language model...")

generator = pipeline(
    "text-generation",
    model="gpt2"
)

print("Model loaded successfully!")

# Step 4: Define the input prompt
prompt = "Artificial Intelligence is transforming the world by"

print("\nInput Prompt:")
print(prompt)

# Step 5: Generate three text responses
print("\nGenerating text responses...")

responses = generator(
    prompt,
    max_new_tokens=60,
    num_return_sequences=3,
    do_sample=True,
    temperature=0.7,
    top_k=50,
    pad_token_id=50256
)

# Step 6: Initialize sentiment analyzer
analyzer = SentimentIntensityAnalyzer()

# Step 7: Analyze the generated responses
analysis_results = []

for i, response in enumerate(responses, start=1):

    generated_text = response["generated_text"]

    # Calculate text statistics
    word_count = len(generated_text.split())
    character_count = len(generated_text)
    sentence_count = sum(
        generated_text.count(punctuation)
        for punctuation in [".", "!", "?"]
    )

    # Perform sentiment analysis
    sentiment_scores = analyzer.polarity_scores(generated_text)
    compound_score = sentiment_scores["compound"]

    if compound_score >= 0.05:
        sentiment = "Positive"
    elif compound_score <= -0.05:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    # Store results
    analysis_results.append({
        "Response Number": i,
        "Generated Text": generated_text,
        "Word Count": word_count,
        "Character Count": character_count,
        "Sentence Count": sentence_count,
        "Sentiment": sentiment,
        "Compound Score": compound_score
    })

# Step 8: Display each generated response
print("\n" + "=" * 60)
print("GENERATED RESPONSES")
print("=" * 60)

for result in analysis_results:
    print(f"\nResponse {result['Response Number']}:")
    print(result["Generated Text"])
    print("Word Count:", result["Word Count"])
    print("Character Count:", result["Character Count"])
    print("Sentence Count:", result["Sentence Count"])
    print("Sentiment:", result["Sentiment"])
    print("Compound Score:", result["Compound Score"])

# Step 9: Display analysis in a table
df = pd.DataFrame(analysis_results)

print("\n" + "=" * 60)
print("COMPARATIVE RESPONSE ANALYSIS")
print("=" * 60)

display(df)

# Step 10: Display overall analysis
print("\n" + "=" * 60)
print("OVERALL ANALYSIS")
print("=" * 60)

print("Total Responses Generated:", len(analysis_results))

average_words = df["Word Count"].mean()
average_characters = df["Character Count"].mean()

print("Average Word Count:", round(average_words, 2))
print("Average Character Count:", round(average_characters, 2))

print("\nSentiment Distribution:")
print(df["Sentiment"].value_counts())

print("\nExperiment completed successfully!")
```
