```python
from transformers import AutoTokenizer

text = "I am learning Natural Language Processing."

bert = AutoTokenizer.from_pretrained("bert-base-uncased")
gpt = AutoTokenizer.from_pretrained("openai-community/gpt2")

bert_tokens = bert.tokenize(text)
gpt_tokens = gpt.tokenize(text)

print("Original Text:", text)

print("\nBERT Tokenization:")
print("Tokens:", bert_tokens)
print("Token IDs:", bert.convert_tokens_to_ids(bert_tokens))

print("\nGPT Tokenization:")
print("Tokens:", gpt_tokens)
print("Token IDs:", gpt.convert_tokens_to_ids(gpt_tokens))

print("\nComparison:")
print("BERT number of tokens:", len(bert_tokens))
print("GPT number of tokens:", len(gpt_tokens))
```
